```python
import pandas as pd

# --- load raw data ---
mh_raw = pd.read_csv("../data/raw/Mental health Depression disorder Data.csv")
gdp_raw = pd.read_csv("../data/raw/gdp-per-capita.csv")

# --- select and rename relevant columns in mental health data ---
mh = mh_raw[['Entity', 'Code', 'Year', 'Depression (%)']].copy()
mh = mh.rename(columns={
    'Entity': 'country',
    'Year': 'year',
    'Depression (%)': 'depression_rate'
})

# convert to numeric
mh['year'] = pd.to_numeric(mh['year'], errors='coerce')
mh['depression_rate'] = pd.to_numeric(mh['depression_rate'], errors='coerce')

# keep only valid rows
mh = mh.dropna(subset=['year', 'depression_rate'])

# restrict to recent years (optional)
mh = mh[mh['year'] >= 1990]

print("Mental health (clean) shape:", mh.shape)

# --- select and rename relevant columns in GDP data ---
gdp = gdp_raw[['Entity', 'Code', 'Year', 'GDP per capita']].copy()
gdp = gdp.rename(columns={
    'Entity': 'country',
    'Year': 'year',
    'GDP per capita': 'gdp_per_capita'
})

gdp['year'] = pd.to_numeric(gdp['year'], errors='coerce')
gdp['gdp_per_capita'] = pd.to_numeric(gdp['gdp_per_capita'], errors='coerce')

gdp = gdp.dropna(subset=['year', 'gdp_per_capita'])

print("GDP (clean) shape:", gdp.shape)

# --- merge on country, Code, year ---
df = pd.merge(
    mh,
    gdp,
    on=['country', 'Code', 'year'],
    how='inner'
)

print("Merged dataset shape:", df.shape)
df.head()
```

```python
import pandas as pd

# Load your base dataset
df = pd.read_csv("../data/clean/mental_econ_merged.csv")

# ---------- Load Gini dataset ----------
gini_raw = pd.read_csv("../data/raw/gini_by_country.csv")

# Rename to match our main dataset
gini = gini_raw.rename(columns={
    "country_name": "country",
    "value": "gini_index"
})

# Clean types
gini["country"] = gini["country"].str.strip()
gini["year"] = pd.to_numeric(gini["year"], errors="coerce")
gini["gini_index"] = pd.to_numeric(gini["gini_index"], errors="coerce")

# Drop incomplete rows
gini = gini.dropna(subset=["year", "gini_index"])

print("Cleaned Gini sample:")
print(gini.head())
print("Gini shape:", gini.shape)

# ---------- Merge with your base dataset ----------
df["country"] = df["country"].str.strip()

df_gini = pd.merge(
    df,
    gini[["country", "year", "gini_index"]],
    on=["country", "year"],
    how="left"
)

print("After merging Gini:", df_gini.shape)
print("Missing Gini values:", df_gini["gini_index"].isna().sum())
df_gini.head()

```

```python
import pandas as pd

# load inequality dataset
ineq_raw = pd.read_csv("../data/raw/global_income_inequality.csv")
print("Inequality columns:", ineq_raw.columns.tolist())
```

```python
# rename columns to nicer / consistent names
ineq = ineq_raw.rename(columns={
    "Country": "country",
    "Year": "year",
    "Population": "population",
    "Gini Index": "gini_index_owid",
    "Average Income (USD)": "avg_income_usd",
    "Top 10% Income Share (%)": "top10_share",
    "Bottom 10% Income Share (%)": "bottom10_share",
    "Income Group": "income_group"
})

# strip spaces from country
ineq["country"] = ineq["country"].astype(str).str.strip()

# numeric conversions
numeric_cols = ["year", "population", "gini_index_owid",
                "avg_income_usd", "top10_share", "bottom10_share"]

for col in numeric_cols:
    ineq[col] = pd.to_numeric(ineq[col], errors="coerce")

# drop rows without year
ineq = ineq.dropna(subset=["year"])

print("Cleaned inequality sample:")
print(ineq.head())
print("Inequality shape:", ineq.shape)
```

```python
# make sure country strings line up
df_gini["country"] = df_gini["country"].astype(str).str.strip()

df_full = pd.merge(
    df_gini,
    ineq[["country", "year", "population", "gini_index_owid",
          "avg_income_usd", "top10_share", "bottom10_share", "income_group"]],
    on=["country", "year"],
    how="left"
)

print("Final enriched dataset shape:", df_full.shape)
print("Missing gini_index:", df_full["gini_index"].isna().sum())
print("Missing gini_index_owid:", df_full["gini_index_owid"].isna().sum())
df_full.head()
```

```python
df_full.to_csv("../data/clean/mental_econ_inequality_full.csv", index=False)
```

```python
import numpy as np

# choose numeric features for ML (drop label + non-numeric)
feature_cols = [
    "year",
    "gdp_per_capita",
    "gini_index",
    "gini_index_owid",
    "population",
    "avg_income_usd",
    "top10_share",
    "bottom10_share",
]

df_ml = df_full.dropna(subset=feature_cols + ["depression_rate"]).copy()

X = df_ml[feature_cols].values
y = df_ml["depression_rate"].values

np.save("../data/clean/X_inequality.npy", X)
np.save("../data/clean/y_depression.npy", y)
np.save("../data/clean/X_inequality_columns.npy", np.array(feature_cols))
```

```python
import statsmodels.formula.api as smf

df_reg = df_ml.copy()

model = smf.ols(
    "depression_rate ~ gdp_per_capita + gini_index + gini_index_owid + "
    "avg_income_usd + top10_share + bottom10_share",
    data=df_reg
).fit()

print(model.summary())
```
