```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("../data/clean/mental_econ_merged.csv")

# ---- 1. Distribution of depression_rate ----
plt.figure(figsize=(8,5))
sns.histplot(df['depression_rate'], bins=30, kde=True)
plt.title("Distribution of Depression Prevalence (%)")
plt.xlabel("Depression (%)")
plt.show()

# ---- 2. Distribution of GDP per capita ----
plt.figure(figsize=(8,5))
sns.histplot(df['gdp_per_capita'], bins=40, kde=True)
plt.title("Distribution of GDP per Capita")
plt.xlabel("GDP per capita (USD)")
plt.xscale("log")  # very important!
plt.show()

# ---- 3. Scatter: GDP vs Depression ----
plt.figure(figsize=(8,5))
sns.scatterplot(data=df, x="gdp_per_capita", y="depression_rate", alpha=0.4)
plt.xscale("log")  # log scale makes pattern visible
plt.title("GDP per Capita vs Depression Rate")
plt.xlabel("GDP per capita (log scale)")
plt.ylabel("Depression (%)")
plt.show()

# ---- 4. Correlation heatmap ----
corr = df[['depression_rate','gdp_per_capita']].corr()
sns.heatmap(corr, annot=True, cmap="Blues", fmt=".2f")
plt.title("Correlation Matrix")
plt.show()
```

```python
df = pd.read_csv("../data/clean/mental_econ_inequality_full.csv")
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10,5))
sns.histplot(df["gini_index"], bins=30, kde=True, color="steelblue")
plt.title("Distribution of Gini Index (Dataset 1)")
plt.xlabel("Gini Index")
plt.show()

plt.figure(figsize=(10,5))
sns.histplot(df["gini_index_owid"], bins=30, kde=True, color="darkred")
plt.title("Distribution of Gini Index (OWID)")
plt.xlabel("Gini Index (OWID)")
plt.show()
```

```python
corr_vars = [
    "depression_rate",
    "gdp_per_capita",
    "gini_index",
    "gini_index_owid",
    "avg_income_usd",
    "top10_share",
    "bottom10_share"
]

plt.figure(figsize=(10,8))
sns.heatmap(
    df[corr_vars].corr(),
    annot=True,
    cmap="Blues",
    vmin=-1,
    vmax=1
)
plt.title("Correlation Heatmap (Inequality + GDP + Depression)")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x=df["gini_index"],
    y=df["depression_rate"],
    alpha=0.5
)
plt.title("Gini Index vs Depression Rate")
plt.xlabel("Gini Index")
plt.ylabel("Depression Rate (%)")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(
    x=df["gini_index_owid"],
    y=df["depression_rate"],
    alpha=0.5,
    color="crimson"
)
plt.title("Gini Index (OWID) vs Depression Rate")
plt.xlabel("Gini Index (OWID)")
plt.ylabel("Depression Rate (%)")
plt.show()
```
